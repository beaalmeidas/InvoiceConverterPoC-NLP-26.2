````python
from google.colab import files

uploaded = files.upload()
markdown = uploaded["img_1.md"].decode("utf-8")
````

````python
import json
import re
from pathlib import Path

import ollama

MODEL = "qwen2.5vl:7b"
OPTIONS = {"temperature": 0, "num_ctx": 16384, "num_predict": 8192}
````

````python
!ollama pull {MODEL}
````

````python
PROMPT = """Return only a complete HTML document with simple CSS and closed tags, no Markdown.
Preserve all recognized text exactly and in order; never correct, omit or invent content.
Keep labels with values and the relative layout of fields and sections.
Preserve every table, row, td/th, rowspan and colspan; do not merge unrelated cells.
"""
cliente = ollama.Client()


def gerar_html(conteudo, formato):
    origem = ("Input: PaddleOCR Markdown with HTML tables." if formato == "markdown" else
              "Input: OCR JSON blocks. Use content as text/table markup, type as label, bbox for layout only; hide coordinates.")
    inicio = time.monotonic()
    resposta = cliente.chat(
        model=MODEL,
        messages=[{"role": "user", "content": f"{PROMPT}{origem}\n\n{conteudo}"}],
        options=OPTIONS,
        keep_alive="10m",
    )
    if resposta.get("done_reason") == "length":
        raise ValueError("HTML truncado: aumente num_predict e num_ctx.")
    print(f"Qwen ({formato}): {time.monotonic() - inicio:.2f} s")
    return re.sub(r"^```(?:html)?\s*|\s*```$", "", resposta["message"]["content"].strip(), flags=re.IGNORECASE)


html = gerar_html(markdown, "markdown")
Path("img_1_rep.html").write_text(html, encoding="utf-8")
````

````python
uploaded = files.upload()
nome = next(iter(uploaded))
data = json.loads(uploaded[nome])

if isinstance(data, dict):
    data = data.get("res", data)
    blocos = [{"type": bloco.get("block_label"), "bbox": bloco.get("block_bbox"),
               "content": bloco.get("block_content")}
              for bloco in data["parsing_res_list"]]
else:
    blocos = data

conteudo = json.dumps(blocos, ensure_ascii=False, separators=(",", ":"))
html = gerar_html(conteudo, "json")
output = Path("nota_fiscal_json.html")
output.write_text(html, encoding="utf-8")
print(f"HTML salvo em {output}")
files.download(str(output))
````
